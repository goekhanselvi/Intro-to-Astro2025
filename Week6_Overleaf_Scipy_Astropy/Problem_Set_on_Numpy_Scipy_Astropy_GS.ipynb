## Question-1

Write a Simple python function to calculate **Escape Velocity** of any celestrial body. Calculate the value for earth.

### Hints
<details>
<summary>Forget what is Escape veloctiy?</summary>
It is the minimum velocity needed to escape from gravitational binding.
It is given by $$v = \sqrt{\frac{2GM}{R}}$$
</details>

````python
#Your Answer
#def v(M,R):
````

````python
import math

def escape_velocity(mass, radius):
    # Universal gravitational constant in m^3/(kg·s^2)
    G = 6.674 * 10**-11
    # Calculate escape velocity using the formula: v_e = sqrt(2GM/r)
    v_e = math.sqrt(2 * G * mass / radius)
    return v_e

# Mass of Earth in kilograms
mass_earth = 5.972 * 10**24
# Radius of Earth in meters
radius_earth = 6.371 * 10**6

# Calculate the escape velocity for Earth
escape_velocity_earth = escape_velocity(mass_earth, radius_earth)
# Print the escape velocity in meters per second
print(f"Escape velocity of Earth: {escape_velocity_earth:.2f} m/s")
````

````python
def rms_velocity(temperature, mol_mass):
    k = 1.38e-23  # Boltzmann constant in J/K
    m = mol_mass / 6.022e23  # Convert g/mol to kg
    v_rms = math.sqrt((3 * k * temperature) / m)
    return v_rms

# Constants
temperature = 300  # Temperature in Kelvin
mol_mass_oxygen = 32.00  # Molar mass of O2 in g/mol

# Calculate RMS velocity of O2
v_rms_oxygen = rms_velocity(temperature, mol_mass_oxygen)

# Assuming escape velocity for Earth is about 11,186 m/s
escape_velocity_earth = 11186

# Compare RMS velocity with escape velocity
print(f"RMS velocity of O2: {v_rms_oxygen:.2f} m/s")
if v_rms_oxygen < escape_velocity_earth:
    print("Oxygen molecules cannot escape Earth's atmosphere.")
else:
    print("Oxygen molecules can escape Earth's atmosphere.")
````

## Question-2

Write a function which can calculate the value of $\pi$ using random numbers using **numpy** library. Try using 1000 sample

````python
#Write your code
````

````python
def estimate_pi(num_samples=1000):
    # Random points in the unit square
    x = np.random.rand(num_samples)
    y = np.random.rand(num_samples)

    # Calculate the distance from the origin
    inside_circle = (x**2 + y**2) <= 1

    # Estimate pi using the ratio of points inside the circle to total points
    pi_estimate = (np.sum(inside_circle) / num_samples) * 4
    return pi_estimate

# Call the function and print the estimated value of pi
pi_value = estimate_pi()
print(f"Estimated value of π using {1000} samples: {pi_value:.4f}")
````

## Question-5

Use Newton's Gravitational Law along with Newton's 2nd law of motion to write the differential equation which earth will follow due to Sun's gravity (sun is fixed in it's place).

1. Now use Scipy to solve the equation. The constants needed must be imported from Astropy.
2. Plot your solution , i.e., x-y plot.

````python
#solution
````

````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
from astropy.constants import G, M_sun
from astropy import units as u

# Define the gravitational constant and the mass of the Sun
G = G.value  # Gravitational constant in m^3 kg^-1 s^-2
M = M_sun.value  # Mass of the Sun in kg

# Define the differential equations
def equations(t, y):
    x, y_pos, vx, vy = y
    r = np.sqrt(x**2 + y_pos**2)
    ax = -G * M * x / r**3  # Acceleration in x
    ay = -G * M * y_pos / r**3  # Acceleration in y
    return [vx, vy, ax, ay]

# Initial conditions: position (x0, y0) and velocity (vx0, vy0)
x0 = 1.496e11  # Initial position of Earth (1 AU in meters)
y0 = 0
vx0 = 0  # Initial velocity in x
vy0 = 29780  # Initial velocity in y (in m/s)

# Time span for the simulation
t_span = (0, 3.154e7)  # One year in seconds
initial_conditions = [x0, y0, vx0, vy0]

# Solve the differential equations
solution = solve_ivp(equations, t_span, initial_conditions, t_eval=np.linspace(0, 3.154e7, 1000))

# Extract the results
x = solution.y[0]
y = solution.y[1]

# Plot the trajectory
plt.figure(figsize=(8, 8))
plt.plot(x, y)
plt.title("Earth's Orbit Around the Sun")
plt.xlabel("X Position (m)")
plt.ylabel("Y Position (m)")
plt.axis('equal')
plt.grid()
plt.show()
````

## Question-6

**NASA Cosmic Background Explorer** (COBE) satellite carried an instrument, **FIRAS** (Far-Infrared Absolute Spectrophotometer) to measure the cosmic microwave background (CMB) radiation, which was confirmed to be distributed according to a black-body curve in accordance with the big bang theory:
$$ I(\nu, T) = \frac{2 h \nu^3 c^2}{\exp(\frac{hc\nu}{k_{b}T})-1}$$
where where the radiation frequency is expressed in wavenumbers, $cm^{-1}$, and the speed of light, c is taken to be in $cm-s^{-1}$.

The data file is cmb_data.txt, which contains measured $I(\nu)$ based on the FIRAS observations. Use scipy `curve_fit` to determine $T$, i.e., the Temperature parameter, along with error.

Note: In the file $I$ is in $erg\cdot s^{-1}\cdot cm^{-1}\cdot sr^{-1}$. Take the estimated $\sigma$ error in the measurement to be $2\times 10^{-6}$$erg\cdot s^{-1}\cdot cm^{-1}\cdot sr^{-1}$.

````python
import numpy as np
from astropy import constants as const
from astropy import units as u
from scipy.optimize import curve_fit

# Fiziksel sabitleri birimlerine çevir
h = const.h.to('erg.s').value  # Planck sabiti
k = const.k_B.to('erg/K').value  # Boltzmann sabiti
c = const.c.to('cm/s').value  # Işık hızı

# Verileri yükle
data = np.loadtxt('cmb_data.txt')
u = data[:, 0]  # Frekanslar
i = data[:, 1] * 1e-6  # Intensiteler (MJy/sr'dan erg/cm²/sr/Hz'ye çevir)

# Planck fonksiyonu
def I(u, T):
    n = 2 * h * c**2 * u**3
    d = (h * c * u) / (k * T)
    return n / (np.exp(d) - 1)

# Eğri uyumu
popt, pcov = curve_fit(I, u, i, p0=[2.7])  # Başlangıç tahmini 2.7 K

# Sonuçları hesapla
T_fit = popt[0]
T_err = np.sqrt(np.diag(pcov))[0]

# Sonuçları yazdır
print(f"Fitted Temperature: {T_fit:.4f} K ± {T_err:.4f} K")
````

## Question-7

Calculate the rest mass energy of a Proton in both joule and MeV.

If you want to get the list of constants present in AstroPy. Check the bottom of this link:https://docs.astropy.org/en/stable/constants/index.html

````python
#solution
````

````python
import astropy.constants as const

# Get the mass of a proton
proton_mass_kg = const.m_p.value  # Mass in kg

# Speed of light
c = const.c.value  # Speed of light in m/s

# Calculate the rest mass energy in joules
rest_mass_energy_joules = proton_mass_kg * c**2

# Convert from joules to MeV
joules_to_mev = 1 / (1.6e-13)  # 1 MeV = 1.6e-13 Joules
rest_mass_energy_mev = rest_mass_energy_joules * joules_to_mev

# Print the results
print(f"Rest mass energy of a proton: {rest_mass_energy_joules:.2e} J")
print(f"Rest mass energy of a proton: {rest_mass_energy_mev:.2f} MeV")
````

### Theory needed for problem-8

**Diurnal motion** is an astronomical term referring to the apparent motion of celestial objects (e.g. the Sun and stars) around Earth, or more precisely around the two celestial poles, over the course of one day.

It is caused by Earth's rotation around its axis, so almost every star appears to follow a circular arc path, called the diurnal circle,often depicted in star trail photography.

So, From the viewpoint of an observer on Earth, the apparent motion of an object on the celestrial sphere follows an arc above the horizon, which is called **diurnal arc**.

![Diurnal arc](https://webapps-cdn.esri.com/CDN/gis-dictionary/Media/diurnal-arc.gif)

Here, Diurnal Arc of a Star moving around the celestrial sphere(red line) in the horizontal system of an observer at lattitude $\phi$. Since, the equatorial plane is inclined by the angle $90^{\circ} - \phi$ against the horizontal plane, the upper culmination of the star at the meridian is given by $a_{max} = 90^{\circ} - \phi + \delta$, where $\delta$ is the declination. The star **rises** at hour angle $h_{rise}$, reaches it's highest altitude when it crosses the meridian at $h=0$ and sets at the horizon at $h_{set} = -h_{rise}$. The value can be given by,
$$ \cos(h_{rise}) = -\tan(\delta) \tan(\phi) $$

**Sidereal Time** is the time for which the star is visible on sky. It is given by $T = 2h_{set}$.

**Problem-8**: Find how long the star `Betelgeuse` is present on sky from my location (Jadavpur, Kolkata, India).

````python
#Solution
````

````python
from astropy.coordinates import EarthLocation, AltAz, get_sun, get_moon
from astropy.time import Time
from astropy.coordinates import SkyCoord
import astropy.units as u
import numpy as np

# Betelgeuse's coordinates
betelgeuse = SkyCoord(ra=88.792939 * u.deg, dec=7.407064 * u.deg)

# Location of Jadavpur, Kolkata
jadavpur = EarthLocation(lat=22.4970 * u.deg, lon=88.3697 * u.deg, height=0 * u.m)

# Time range
time_start = Time("2025-08-01")
time_end = Time("2025-08-31")
times = time_start + (time_end - time_start) * np.linspace(0, 1, 100)

# Alt-Az coordinates
altaz_frame = AltAz(obstime=times, location=jadavpur)
betelgeuse_altaz = betelgeuse.transform_to(altaz_frame)

# Find rise and set times of Betelgeuse
rise_times = times[betelgeuse_altaz.alt > 0]
set_times = times[betelgeuse_altaz.alt < 0]

# Calculate rise and set times
if len(rise_times) > 0 and len(set_times) > 0:
    rise_time = rise_times[0]
    set_time = set_times[-1]
    duration = set_time - rise_time
    print(f"Betelgeuse rises at {rise_time.iso} and sets at {set_time.iso} from Jadavpur, Kolkata.")
    print(f"Duration the star is visible: {duration}")
else:
    print("Betelgeuse is not visible during these dates.")
````

````python
from astropy.coordinates import EarthLocation, AltAz, SkyCoord
from astropy.time import Time
import astropy.units as u
import numpy as np

# Betelgeuse'nin koordinatları
betelgeuse = SkyCoord(ra=88.792939 * u.deg, dec=7.407064 * u.deg)

# Jadavpur, Kolkata konumu
jadavpur = EarthLocation(lat=22.4970 * u.deg, lon=88.3697 * u.deg, height=0 * u.m)

# Zaman aralığı
time_start = Time("2025-08-01")
time_end = Time("2025-08-31")
times = time_start + (time_end - time_start) * np.linspace(0, 1, 100)

# Alt-Az koordinatları
altaz_frame = AltAz(obstime=times, location=jadavpur)
betelgeuse_altaz = betelgeuse.transform_to(altaz_frame)

# Betelgeuse'nin yükselme ve batma zamanlarını bulma
rise_times = times[betelgeuse_altaz.alt > 0]
set_times = times[betelgeuse_altaz.alt < 0]

# Yükselme ve batma zamanlarını hesapla
if len(rise_times) > 0 and len(set_times) > 0:
    rise_time = rise_times[0]
    set_time = set_times[-1]
    duration = set_time - rise_time
    print(f"Betelgeuse rises at {rise_time.iso} and sets at {set_time.iso} from Jadavpur, Kolkata.")
    print(f"Duration the star is visible: {duration}")
else:
    print("Betelgeuse is not visible during these dates.")
````

## Question-9

The **Declination** of sun $\delta_s$ is given by,
$$ \delta_s = -\arcsin\Bigg(\sin(\epsilon_0)\cos\Big(\frac{360}{365.24}(N+10)\Big)\Bigg)$$
where $\epsilon_0 = 23.44^{\circ}$ and $N$ is the difference in days starting from 1st january.

Make a plot of how the length of day changes over the year in your location.

````python
# Constants
days_in_year = 365.24
declination_angle = 23.44  # in degrees
N = np.arange(1, days_in_year + 1)

# Calculate declination
declination = -np.arcsin(np.sin(np.radians(declination_angle)) *
                         np.cos((360 / days_in_year) * (N + 10) * (np.pi / 180)))

# Calculate day length (simplified formula for day length)
day_length = 24 * (1 + (declination / (np.pi / 2)))

# Plotting
plt.figure(figsize=(10, 5))
plt.plot(N, day_length, label='Day Length (hours)')
plt.title('Variation of Day Length Throughout the Year')
plt.xlabel('Day of the Year')
plt.ylabel('Length of Day (hours)')
plt.xticks(np.arange(0, 366, 30))  # Show every month
plt.grid()
plt.legend()
plt.show()
````
