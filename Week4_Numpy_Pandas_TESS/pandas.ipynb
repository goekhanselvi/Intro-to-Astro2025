```python
import pandas as pd
```

```python
df=pd.read_csv('titanic.csv')
```

```python
df.head()
```

```python
df['Title'] = df['Name'].str.extract(r' ([A-Za-z]+)\.', expand=False)
```

```python
title_survival_rate = df.groupby('Title')['Survived'].mean()
print(title_survival_rate)
```

```python
df['CabinGroup'] = df['Cabin'].str[0]
df['CabinGroup'] = df['CabinGroup'].fillna('Unknown')

cabin_stats = df.groupby('CabinGroup').agg(
    Total_Passengers=('PassengerId', 'count'),
    Survival_Rate=('Survived', 'mean')
)
print(cabin_stats)
```

```python
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1

def categorize_family(size):
    if size == 1:
        return 'Single'
    elif 2 <= size <= 4:
        return 'Small'
    else:
        return 'Large'

df['FamilyCategory'] = df['FamilySize'].apply(categorize_family)

family_survival = df.groupby('FamilyCategory')['Survived'].mean()
print(family_survival)
```

```python
bins = [0, 10, 20, 30, 40, 50, 60, 70, 120]
labels = ['0–10', '11–20', '21–30', '31–40', '41–50', '51–60', '61–70', '71+']

df['AgeGroup'] = pd.cut(df['Age'], bins=bins, labels=labels)

age_group_survival = df.groupby('AgeGroup')['Survived'].mean()
print(age_group_survival)

import matplotlib.pyplot as plt

age_group_survival.plot(kind='bar', title='Survival Rate by Age Group')
plt.ylabel('Survival Rate')
plt.xlabel('Age Group')
plt.show()
```

```python
pivot = pd.pivot_table(df, values='Survived', index='Pclass', columns='Sex', aggfunc=['sum', 'mean'])
print(pivot)
```

```python
title_medians = df.groupby('Title')['Age'].median()

df['Age'] = df.apply(lambda row: title_medians[row['Title']] if pd.isnull(row['Age']) else row['Age'], axis=1)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

df_encoded = df.copy()

df_encoded['Sex'] = df_encoded['Sex'].map({'male': 0, 'female': 1})
df_encoded['Embarked'] = df_encoded['Embarked'].map({'S': 0, 'C': 1, 'Q': 2})

df_encoded = pd.get_dummies(df_encoded, columns=['Title'], drop_first=True)

numeric_df = df_encoded.select_dtypes(include=['number'])

corr_matrix = numeric_df.corr()

plt.figure(figsize=(12, 8))
sns.heatmap(corr_matrix[['Survived']].sort_values(by='Survived', ascending=False), annot=True, cmap='coolwarm')
plt.title("Correlation with Survived")
plt.show()
```

```python
ticket_counts = df['Ticket'].value_counts()
duplicate_tickets = ticket_counts[ticket_counts > 1].index

shared_ticket_survival = df[df['Ticket'].isin(duplicate_tickets)].groupby('Ticket')['Survived'].mean()
print(shared_ticket_survival.head())
```

```python
df['FarePerPerson'] = df['Fare'] / df['FamilySize']

correlation = df['FarePerPerson'].corr(df['Survived'])
print("FarePerPerson ile Survived korelasyonu:", correlation)
```

```python
df['FareBin'] = pd.qcut(df['Fare'], 4, labels=['Low', 'Mid', 'High', 'VeryHigh'])

df['PclassFareGroup'] = df['Pclass'].astype(str) + '_' + df['FareBin'].astype(str)

fare_group_survival = df.groupby('PclassFareGroup')['Survived'].mean()
print(fare_group_survival)
```
